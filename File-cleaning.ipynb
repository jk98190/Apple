```python
#Import the requried libaraies 
import pandas as pd 
import numpy as np
```

```python
#Read the files 
financial_facts=pd.read_csv("financial_facts.csv")
filings=pd.read_csv("filings.csv")
```

```python
# Cleaning file inspection for file 1 
print("== Messy data file to csv==")
financial_facts.columns = (financial_facts.columns.str.strip().str.lower()
               .str.replace(" ", "_").str.replace(r"[^\w]", "", regex=True))

print(financial_facts.shape,financial_facts.dtypes, sep="\n")
print("Missing %:\n", (financial_facts.isna().mean() * 100).loc[lambda s: s > 0].sort_values(ascending=False))

# Dedupe, strip text
financial_facts=financial_facts.drop_duplicates()
obj =financial_facts.select_dtypes("object").columns
financial_facts[obj]=financial_facts[obj].apply(lambda s: s.str.strip())

# df1["date"] = pd.to_datetime(df1["date"], errors="coerce")  # uncomment per column
# df1["sales"] = pd.to_numeric(df1["sales"], errors="coerce")

# Fill missing: median for numeric, mode for text
num = financial_facts.select_dtypes("number").columns
financial_facts[num] =financial_facts[num].fillna(financial_facts[num].median())
financial_facts[obj] =financial_facts[obj].fillna(financial_facts[obj].mode().iloc[0])

# IQR outlier counts (report only, not removed)
q1, q3 =financial_facts[num].quantile(.25),financial_facts[num].quantile(.75)
iqr = q3 - q1
print("Outliers:\n", ((financial_facts[num] < q1 - 1.5*iqr) | (financial_facts[num] > q3 + 1.5*iqr)).sum())

print("Missing left:",financial_facts.isna().sum().sum(), "| Dupes:",financial_facts.duplicated().sum(),"| Shape:",financial_facts.shape)
financial_facts.to_csv("cleaned_data.csv", index=False)
```
